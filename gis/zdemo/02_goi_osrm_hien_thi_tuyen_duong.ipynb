```python
# Nạp thư viện cần dùng để gọi OSRM và giải mã tuyến đường
import requests
import polyline
```

```python
# Khai báo địa chỉ API OSRM
# API này dùng để lấy tuyến đường đẹp phục vụ hiển thị trên bản đồ
OSRM_BASE_URL = "http://router.project-osrm.org/route/v1"
```

```python
# Hàm lấy route_points từ OSRM
def get_osrm_route_points(start_lat, start_lng, end_lat, end_lng, delivery_mode='motorbike'):
    osrm_profile = 'foot' if delivery_mode == 'walking' else 'driving'
    route_coordinates = f"{start_lng},{start_lat};{end_lng},{end_lat}"
    request_url = f"{OSRM_BASE_URL}/{osrm_profile}/{route_coordinates}"

    query_params = {
        'overview': 'full',
        'geometries': 'polyline',
        'steps': 'true',
        'alternatives': 'false',
    }

    request_headers = {
        'User-Agent': 'Mozilla/5.0',
    }

    response = requests.get(
        request_url,
        params=query_params,
        headers=request_headers,
        timeout=10,
    )

    response_data = response.json()

    if response.status_code == 200 and response_data.get('code') == 'Ok':
        route_list = response_data.get('routes', [])
        if route_list:
            encoded_geometry = route_list[0].get('geometry')
            if encoded_geometry:
                return polyline.decode(encoded_geometry)

    return [
        [float(start_lat), float(start_lng)],
        [float(end_lat), float(end_lng)],
    ]
```

```python
#Nạp dữ liệu đầu vào
# Đây là tọa độ mẫu để gọi thử OSRM
start_lat = 10.7721
start_lng = 106.6983
end_lat = 10.7765
end_lng = 106.7009
delivery_mode = 'motorbike'
```

```python
#Gọi hàm lấy route_points từ OSRM

route_points = get_osrm_route_points(
    start_lat=start_lat,
    start_lng=start_lng,
    end_lat=end_lat,
    end_lng=end_lng,
    delivery_mode=delivery_mode,
)

print("Số điểm của tuyến đường:", len(route_points))
print("5 điểm đầu của tuyến đường:")
print(route_points[:5])
```
